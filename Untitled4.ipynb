```python
# Autor: Estudiante de Ciencia de Datos
# Actividad 05: Integrar sistema con restricciones (Ordenamiento por Selección)

def selection_sort(arr):
    """Ordena un arreglo de forma ascendente usando el algoritmo Selection Sort."""
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        # Intercambiamos el elemento menor encontrado con el primero de la iteración
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# Dataset desorganizado de temperaturas de la estación meteorológica
temperaturas = [25.4, 12.8, 35.1, 4.2, 18.5]

print("Temperaturas originales:", temperaturas)
temperaturas_ordenadas = selection_sort(temperaturas)
print("Temperaturas ordenadas:", temperaturas_ordenadas)
```
